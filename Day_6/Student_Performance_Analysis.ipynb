```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("C:\\Users\\92348\\Desktop\\MLBench\\Day_6\\student_performance.csv")
df = pd.DataFrame(data) 
```

```python
#Loop 
#here i am showing we can do via loop and built-in functions also
python_avg = df['Python']
total = 0
count = 0
for i in python_avg:
    total += i
    count += 1
if count > 0:
    mean = total / count
    print(f"Mean of Python scores: {mean}")
else:
    print("No valid scores found in Python column")
```

```python
# df[['Python', 'Mathematics', 'Statistics', 'Machine_Learning']].mean()
python_avg = df['Python'].mean()
math_avg = df['Mathematics'].mean()
stats_avg = df['Statistics'].mean()
ml_avg = df['Machine_Learning'].mean()
print(f"Mean of Python scores: {python_avg}")
print(f"Mean of Mathematics scores: {math_avg}")
print(f"Mean of Statistics scores: {stats_avg}")
print(f"Mean of Machine Learning scores: {ml_avg}")
```

```python
top_students = df.nlargest(5, ['Python', 'Mathematics', 'Statistics', 'Machine_Learning'])
print("Top 5 students based on performance in all subjects:")
print(top_students[['Name', 'Python', 'Mathematics', 'Statistics', 'Machine_Learning']])
```

```python
#Students in Python Below Average || Mean of Python scores: 78.9
below_python_avg = df[df['Python'] < python_avg]
print(f"Students who are below than average in Python:", below_python_avg[['Name', 'Python']])
```

```python
#Students in Mathematics Below Average || Mean of Mathematics scores: 79.5
below_math_avg = df[df['Mathematics'] < math_avg]
print(f"Students who are below than average in Mathematics:", below_math_avg[['Name', 'Mathematics']])
```

```python
#Students in Statistics Below Average || Mean of Statistics scores: 80.6
below_stats_avg = df[df['Statistics'] < stats_avg]
print(f"Students who are below than average in Statistics:", below_stats_avg[['Name', 'Statistics']])
```

```python
#Students in Machine Learning Below Average || Mean of Machine Learning scores: 82.6
below_ml_avg = df[df['Machine_Learning'] < ml_avg]
print(f"Students who are below than average in Machine Learning:", below_ml_avg[['Name', 'Machine_Learning']])
```
