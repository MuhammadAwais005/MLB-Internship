Task 5
Split an RGB image into R, G, and B channels manually and save each channel as a separate image

```python
import cv2
import numpy as np
```

```python


img = cv2.imread(r"C:\Users\92348\Desktop\MLBench\Enjoying.jpg")  
h, w, _ = img.shape

R = np.zeros((h, w), dtype=np.uint8)
G = np.zeros((h, w), dtype=np.uint8)
B = np.zeros((h, w), dtype=np.uint8)

for i in range(h):
    for j in range(w):
        B[i, j] = img[i, j, 0]
        G[i, j] = img[i, j, 1]
        R[i, j] = img[i, j, 2]
        #0 Blue, 1 Green, 2 Red

cv2.imwrite("red_channel.png", R)
cv2.imwrite("green_channel.png", G)
cv2.imwrite("blue_channel.png", B)
```
