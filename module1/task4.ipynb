```python
image = cv2.imread(r'C:\Users\92348\Desktop\MLBench\Enjoying.jpg')
```

```python
total_pixels =0
black_pixels = 0
white_pixels = 0
intensity = 0

for rows in image:
    for pixels in rows:
        total_pixels += 1
        total_pixels= total_pixels * 3
        if pixels[0] == 0 and pixels[1] == 0 and pixels[2] == 0:
            black_pixels += 1
        elif pixels[0] == 255 and pixels[1] == 255 and pixels[2] == 255:
            white_pixels += 1
        if pixels[0] > 200 and pixels[1] > 200 and pixels[2] > 200:
            intensity += 1

print("Total pixels:", total_pixels)
print("Black pixels:", black_pixels)
print("White pixels:", white_pixels)
print("Pixels with intensity > 200:", intensity)
```
