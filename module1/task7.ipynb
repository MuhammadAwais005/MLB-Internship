Calculate how much memory (in bytes and MB) an image occupies using its dimensions, channels, and bit depth.

```python
import cv2
```

```python


# Load image
img = cv2.imread(r"C:\Users\92348\Desktop\MLBench\Enjoying.jpg")


if img is None:
    print("Error: Could not load image.")
    exit()


height, width = img.shape[:2]


channels = img.shape[2] if len(img.shape) == 3 else 1


bit_depth = img.dtype.itemsize * 8


memory_bits = height * width * channels * bit_depth
memory_bytes = memory_bits // 8
memory_mb = memory_bytes / (1024 ** 2)

# Display information
print("----- Image Information -----")
print(f"Width       : {width} pixels")
print(f"Height      : {height} pixels")
print(f"Channels    : {channels}")
print(f"Data type   : {img.dtype}")
print(f"Bit depth   : {bit_depth} bits")

print("\n----- Memory -----")
print(f"Memory      : {memory_bytes:,} bytes")
print(f"Memory      : {memory_mb:.2f} MB")
```
