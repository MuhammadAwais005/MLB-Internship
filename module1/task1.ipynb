Load an image and print:
Height
Width
Number of channels
Data type
Total number of pixels

```python
import cv2
import numpy as np
```

```python
image = cv2.imread(r'C:\Users\92348\Desktop\MLBench\Enjoying.jpg')
#to check either our image is correctly loaded or not:

if image is None:
    print("Error: Could not load image.")
else:
    print("Image loaded successfully.")
```

```python
print(image.shape)
```

```python
if len(image.shape) > 2:
    print("The image is a color image.")
else:
    print("The image is a grayscale image.")
```

```python
# height, width, channels

img = np.array(image)
height = img.shape[0]  
width = img.shape[1]

if len(img.shape) > 2:
    channels = img.shape[2]
else:
    channels = 1
```

```python
#datatype, total pixels
datatype = img.dtype
pixels = height * width * channels
```

```python
print("Image Information: ")
print(f"Dimensions: {height} x {width}")
print(f"Channels: {channels}")
print(f"Data Type: {datatype}")
print(f"Total Pixels: {height} x {width} x {channels} = {pixels}")
```
