Without using np.mean() or cv2.mean(), calculate:
Mean pixel intensity
Minimum pixel value
Maximum pixel value

```python
import cv2
import numpy as np
```

```python
image = cv2.imread(r'C:\Users\92348\Desktop\MLBench\Enjoying.jpg')
if image is None:
    print("Error: Could not read the image.")
else:
    print("Image read successfully.")
```

```python
img = np.array(image)
```

```python
height = img.shape[0]  
width = img.shape[1]

if len(img.shape) > 2:
    channels = img.shape[2]
else:
    channels = 1
```

```python
total = 0
count = 0

minimum = 255
maximum = 0

# Go through every pixel
for row in img:
    for pixel in row:

        # Go through R, G, B
        for value in pixel:

            total = total + int(value)
            count = count + 1

            if value < minimum:
                minimum = int(value)

            if value > maximum:
                maximum = int(value)

mean = total / count

print("Mean pixel intensity:", mean)
print("Minimum pixel value:", minimum)
print("Maximum pixel value:", maximum)
```
