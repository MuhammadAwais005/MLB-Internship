Convert an RGB image to Grayscale without using cv2.cvtColor().


```python
import numpy as np
import cv2
```

```python
# Load RGB image
img = np.array(cv2.imread(r"C:\Users\92348\Desktop\MLBench\Enjoying.jpg"))

height = img.shape[0]
width = img.shape[1]

# Create empty grayscale image
gray = np.zeros((height, width), dtype=np.uint8)
```

```python

for row in range(height):
    for col in range(width):

        R = img[row, col, 0]
        G = img[row, col, 1]
        B = img[row, col, 2]

        gray_value = 0.299 * R + 0.587 * G + 0.114 * B

        gray[row, col] = int(gray_value)

# Save grayscale image
cv2.imwrite("grayscale.png", gray)
```
